```python
# PyTorch Dataset and DataLoader

## Why mini-batches?
# - Loading the full dataset at once is memory-heavy and slow.
# - Mini-batch gradient descent splits data into small batches, so training is faster and fits in RAM.

## Dataset
# - `Dataset` is responsible for **getting one sample at a time**.
# - Main methods:
#   - `__init__()` → locate/load data
#   - `__len__()` → return total number of samples
#   - `__getitem__(idx)` → return one sample for a given index
# - Best place for sample-level transforms like resize, normalize, etc.

## DataLoader
# - `DataLoader` is responsible for **making batches** from the Dataset.
# - It:
#   1. picks indices
#   2. optionally shuffles them
#   3. calls `Dataset.__getitem__()` for each index
#   4. uses `collate_fn` to combine samples into one batch

## Collate function
# - Takes a list of individual samples and packs them into a batch.
# - Example:
#   - single image: `(3, 224, 224)`
#   - batch of 32 images: `(32, 3, 224, 224)`

## Useful DataLoader options
# - `batch_size` → number of samples per batch
# - `shuffle=True` → randomize sample order
# - `num_workers` → load data in parallel
# - `drop_last=True` → drop the last incomplete batch

## Main idea
# - `Dataset` = fetch data  
# - `DataLoader` = batch and deliver data to the model
```

```python
import numpy as np
import pandas as pd
import torch
from torch.utils.data import DataLoader, Dataset
```

```python
from sklearn.datasets import make_classification


# Creating a dataset
x, y = make_classification(
    n_samples=10,
    n_features=2,
    n_informative=2,
    n_redundant=0,
    n_classes=2,
    random_state=42
)

print(x, "\n")
print(y)
```

```python
# Converting x, y in Tensors
x = torch.tensor(x, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.float)
```

```python
# Creating a Dataset Class

class CustomDataset(Dataset):
    def __init__(self, inputs, outputs):
        self.features = inputs
        self.labels = outputs

    def __len__(self):
        return self.inputs.shape[0]
    
    def __getitem__(self, index):
        return self.inputs[index], self.outputs[index]
```

```python
dataset = CustomDataset(x, y)
```

```python
# Creating DataLoader Object (no need to create class)
dataloader = DataLoader(dataset, batch_size=3, shuffle=True, drop_last=True)
```

```python
for batch_features, batch_labels in dataloader:
    print(batch_features)
    print(batch_labels)
    print("-"*50)
```
