```python
# Autograd is a core component of PyTorch that provides automatic differentiation for tensor operations. 
# It enables gradient computation, which is essential for training machine learning models using optimization
# algorithms like gradient descent.

# Input tensors -> leaf | Output tensors -> root, that's why we always compute .backwards() for root.
```

```python
import torch
import math

x = torch.tensor(3.0, requires_grad=True)
```

```python
y = x ** 2

print(y)
```

```python
x = torch.tensor(4.0, requires_grad=True)
y = x ** 2
z = torch.sin(y)
```

```python
z.backward() # <- this computes the 
```

```python
x.grad # <- this stores the gradient
```

```python
# Grad for Vectors
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)

y = (x**2).mean()
```

```python
y.backward()
```

```python
print(x.grad)
```

```python
# Cleaning Grad
# If we do not clean the grad in the end then all the grad values will accumulate
#  and will be added in the result which can give workng values, to fix this
#  we use clearning grad func:
x.grad.zero_()
```

```python
# Prac Qes
x = torch.tensor(2.0, requires_grad=True)

y = (x**2) + (3*x) + 1

y.backward()

x.grad
```
