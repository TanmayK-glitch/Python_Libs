```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
class SimpleModel(nn.Module):
    def __init__(self, num_features):

        super().__init__()
        self.linear = nn.Linear(num_features, 1) #(input_feat, output_feat)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        out = self.linear(features)
        out = self.sigmoid(out)

        return out
```

```python
# Creating a dataset
features = torch.rand(10, 5)

# Create Model
model = SimpleModel(features.shape[1])

# Call the model for forward pass
print(model(features))
```

```python
class NewSimpleNN(nn.Module):
    def __init__(self, input_feat):
        super().__init__()
        self.linear1 = nn.Linear(input_feat, 3)
        self.relu = nn.ReLU()
        self.linear2 = nn.Linear(3, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        out = self.linear1(features)
        out = self.relu(out)
        out = self.linear2(out)
        out = self.sigmoid(out)

        return out
```

```python
# Creating a DS
features = torch.rand(10, 5)

model = NewSimpleNN(features.shape[1])

model(features)
```

### Instead of passing the out(y_pred) values through all the methods we can use SEQUENTIAL CONTAINERS to do it.

```python
class NewSimpleNN(nn.Module):
    def __init__(self, input_feat):
        super().__init__()

        self.network(
        nn.Linear(input_feat, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
        )

    def forward(self, features):
        out = self.network(features)

        return out
```

## Full Training Pipeline using in built Loss and Optimizer

```python
# Preapre Data
x = torch.rand(100, 1) * 10
y = x * 2
```

```python
# Defining the Module
class TrainingPipeline(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(in_features=1, out_features=1)

    def forward(self, x):
        return self.linear(x)
```

```python
model = TrainingPipeline()
```

```python
# Loss Func
criterion = nn.MSELoss()

# Optimizer
optimizer = optim.SGD(model.parameters(), lr=0.01)

# Setting EPOCH
epoch = 100
```

```python
# Training loop
for epoch in range(epoch):

    # 1. Forward pass
    predictions = model(x)

    # 2. Loss func
    loss = criterion(predictions, y)

    # 3. Gradient Descent
    optimizer.zero_grad()

    # 4. Backward pass
    loss.backward()

    # 5. Update the weights
    optimizer.step()
```

```python
# Testing Model
model.eval()

test_input = torch.tensor([[5.0]])

with torch.no_grad():
    test_prediction = model(test_input)

print(f"Input 5.0: | Model Prediction: {test_prediction.item(): .4f}")
```
