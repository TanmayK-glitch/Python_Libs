```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('matches - matches.csv')
data.head(1)
```

```python
# <------------------------The Pandas Plot()--------------------------->
# These are the major types of Graphs
# 1. Line Plot {
# time-series data
# trends, growth, decline
# }

# 2. Bar Plot {
# comparing categories
# frequencies
#}

# 3. Histogram {
# understanding distribution of numerical data
# }
data['winner'].value_counts().plot(kind = 'barh') #barh - horizontal ploting
```

```python
data['toss_decision'].value_counts().plot(kind = 'pie')
```

```python
data['win_by_runs'].plot(kind = 'hist') #Histogram is mainly used for Numerical Data.
```
