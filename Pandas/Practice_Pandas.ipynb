```python
deliveriesData = pd.read_csv('deliveries.csv')
matchesData = pd.read_csv('matches.csv')
deliveriesData.head(2)
```

```python
#Q2 Total extras conceded by each bowling team
bowlingTeam = deliveriesData.groupby('bowling_team')
bowlingTeam['extra_runs'].count().sort_values(ascending=False).head(10)
```

```python
#Q3 Batsman with the most fours hit
batsmanRuns = deliveriesData['batsman_runs'] == 4
newMask = deliveriesData[batsmanRuns]
newMask.groupby('batsman')['batsman_runs'].count().sort_values(ascending=False).head(5)
```

```python
#Q4 Bowlers who have bowled the most no-balls
bolwerMask = deliveriesData['noball_runs'] > 0
mostNoBalls = deliveriesData[bolwerMask]
mostNoBalls.groupby('bowler')['noball_runs'].sum().sort_values(ascending=False).head(5)
```

```python
#Q5 For batsman 'V Kohli', how many legal deliveries (where ball == 1 in the over) has he faced?
legal_kohli = deliveriesData[(deliveriesData['batsman'] == 'V Kohli') & (deliveriesData['wide_runs'] == 0) & (deliveriesData['noball_runs'] == 0)]
balls_faced = len(legal_kohli)          # or legal_kohli.shape[0]
print(balls_faced)
```

```python
# isin
#Find the top 5 bowlers who have the best (lowest) average in death overs
#(overs 16–20), but only consider bowlers who have bowled at least 30 legal deliveries in death overs.
filter_bowler = deliveriesData['bowler']
death_delivery = filter_bowler[(deliveriesData['over'] >= 16)]
bowler_mask = deliveriesData[death_delivery['noball_runs'] == 0]
```
