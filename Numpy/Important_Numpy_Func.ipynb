```python
import numpy as np
```

```python
# Generates random floating point number bw 0 - 1
n1 = np.random.random(10)
print(n1)
```

```python
# seed() -> gennerate same random number everytime we run the func
n2 = np.random.seed(1)
n2 = np.random.rand()
n2 = np.random.randint(1, 10)
print(n2)
```

```python
#.uniform() -> generates a random floating point number within specific range
n3 = np.random.uniform(1, 10, 10)
print(n3)
```

```python
# Statistical and Search Functions
# 1. max() -> find max number in an array || min() -> finds min number in an array
n4 = np.arange(1, 10)
max_num = np.max(n4)
min_num = np.min(n4)
print(min_num)
print(max_num)
```

```python
# 2. argmax() -> return the index position of max value in the array
max_index = np.argmax(n4)
min_index = np.argmin(n4)
print(max_index)
print(min_index)
```

```python
# 3. percentile() -> it tells "At this percentage point, what data value do it hit?"
n5 = np.arange(1, 11)
print(n5)
per_value = np.percentile(n5, 70)
print(per_value)
```

```python
# Manipulation Functions
# 1. np.where() -> it is used to modify an array based on a specific condition, without changing the original array in place,
# It requires 3 parameters (the condition, value if true, value if false)
n6 = np.arange(1, 10)
mod_arr = np.where(n6 % 2 == 0, -1, 0)
print(mod_arr)
```
