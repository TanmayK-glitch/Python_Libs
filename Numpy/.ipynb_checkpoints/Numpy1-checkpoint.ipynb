```python
#Creating Numpy / n-d Arrays
import numpy as np
```

```python
arr1 = np.array([1, 2, 3, 4, 5])
print(arr1)
```

```python
type(arr1)
```

```python
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(arr2)
```

```python
arr3 = np.zeros((2, 3))
print(arr3)
```

```python
arr4 = np.ones((3, 3))
print(arr4)
```

```python
arr5 = np.identity(5)
print(arr5)
```

```python
arr6 = np.arange(10)
print(arr6)
```

```python
arr7 = arr6.copy()
print(arr7)
```
