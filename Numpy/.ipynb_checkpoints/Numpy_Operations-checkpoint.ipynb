```python
import numpy as np
```

```python
arr1 = np.array([1, 2, 3, 4, 5])
arr2 = np.array([6, 7, 8, 9, 10])
```

```python
# It does element wise operations
arr1 - arr2
```

```python
arr1 * arr2
```

```python
arr1 > 3 #Get the value of each element (true / false)
```

```python
# Dot is used to get the dot product, here order of matrix matters
arr2 = np.arange(6).reshape(2,3)
arr3 = np.arange(6, 12).reshape(3,2)
dot_result = arr2.dot(arr3)
print(dot_result) 
print(arr3)
```

```python
max_num = arr3.max() #Gives max element of the matrix (same with min())
max_axisY = arr3.max(axis=0) #Gives max element from the ROWS
max_axisX = arr3.max(axis=1) #Gives max element from the COLS
print(max_axisY)
print(max_axisX)
print(max_num)
```

```python
# Can use a lot of built in Math functions like -> mean, median, std, sin
sin_val = np.sin(arr3)
med_val = np.median(arr2)
exp_val = np.exp(arr2)
print(exp_val)
print(med_val)
print(sin_val)
```
