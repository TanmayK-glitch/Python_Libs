```python
#List vs Array
#Why is Array (Numpy) faster than List ??
# Array -> Faster, Convenient, and uses Less Memory
import numpy as np
```

```python
listA = range(100)
arrA = np.arange(100)
```

```python
import sys
```

```python
#Comparison of size of Array and List
print(sys.getsizeof(55) * len(listA))
print(arrA.itemsize * arrA.size)
```
