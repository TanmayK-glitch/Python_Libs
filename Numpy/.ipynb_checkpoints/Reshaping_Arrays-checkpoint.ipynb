```python
import numpy as np
```

```python
arr1 = np.arange(6).reshape(2,3)
print(arr1)
```

```python
# Can use .ravel to convert any n-dim array to 1D array
arr1.ndim # Use ndim to get to know the dim of any array
arr1.ravel()
```

```python
#Transpose 
arr1.transpose()
```

```python
#Stacking -> can combine 2 arrays
arr2 = np.arange(1, 7).reshape(2, 3)
arr3 = np.arange(7, 13).reshape(2, 3)
arr4 = np.arange(14, 20).reshape(2,3)
np.hstack((arr2, arr3)) # hstack -> horizontal 
np.vstack((arr2, arr3)) # vstack -> vertical 
```

```python
# Splitting 
print(arr2)
np.hsplit(arr2, 3) # the value input is how much you want to spli it, it should be equal according to the matrix
np.vsplit(arr2, 2)
```

```python
arr5 = np.random.randint(low=1, high=100, size=20).reshape(4, 5)
print(arr5)
```

```python
mask = arr5 > 10
print(mask) # Mask retruns only the values whoes conditions are true
arr5[arr5 > 10] # This is used to filter 
```

```python
arr5[(arr5 > 50) & (arr5%2!=0)] # Here we use bitwise & not the "and"
```

```python
arr5[(arr5 > 50) & (arr5%2!=0)]=0 #Changes the value of those elements who satisfies the condition
print(arr5) 
```
