```python
#Numpy Attributes
import numpy as np
```

```python
arr1 = np.array([[[1,2],[3,4]],
                 [[5,6],[7,8]]])
print(arr1)
```

```python
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(arr2)
```

```python
#Shape is used to get to know the SHAPE of a Matrix
#Read as (layer, rows, columns) -> Above matrix has 2 Matrix (1 beside other) + 2 Rows + 2 Columns
arr1.shape
```

```python
#ndim used to get the dimestion of the Matrix
#ndim is an "Attribute" thats why we do not use any "()" in front for it
arr1.ndim
```

```python
arr2.size #Size -> total elements in the matrix
arr2.itemsize #Itemsize -> total size of the elemtes used, here INT is 8 Bytes (64Bit)
```

```python
#Dtype -> data type with also Bits
arr1.dtype
```
