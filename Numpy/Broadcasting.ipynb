```python
# Broadcasting 
# It is a concept which is used while performing operations on 2 arrays.
#-------------------------------------------------------------------------
# If the size of array is same then the array will perform as usual but,
# if the size is not same then still it performs the operation but it has 
# some conditions:
# 1. The "y" value should be same of both arrays (x, y).
# 2. If x=1 and y=1 then the operation will take place.
```

```python
import numpy as np
```

```python
# Scenario 1
a1 = np.arange(8).reshape(2, 4)
a2 = np.arange(8, 16).reshape(2, 4)
a1 + a2
```

```python
# Scenario 2
a3 = np.arange(9).reshape(3, 3)
a4 = np.arange(3).reshape(1, 3)
a3 + a4
```

```python
# Here the condition(x=1 and y-should be same) does not match, so it wont broadcast the arrays
a5 = np.arange(3).reshape(1, 3)
a6 = np.arange(16).reshape(4, 4)
a5 + a6
```

```python
# Here x=1 and n=1 then y==m, so the operation will take place
a7 = np.arange(3).reshape(1, 3)
a8 = np.arange(3).reshape(3, 1)
a7 + a8
```

```python
# If x=1 and y=1 then operation will take place no matter what 
a9 = np.arange(1).reshape(1, 1)
a10 = np.arange(10).reshape(2, 5)
a9 + a10
```

```python
# Broadcasting even works on differnt dimestions coz the y is same for both the arrays
a11 = np.arange(4)
a12 = np.arange(20).reshape(5, 4)
a11 + a12
```
